Repeat VS On-time Customers

```python
customer_order = data.groupby('customer_unique_id')['order_id'].nunique()

customer_order
```

```python
repeat_customer = (customer_order>1).sum()
one_time_customer = (customer_order == 1).sum()


print("Repeat:", repeat_customer)
print("one_time_customer:", one_time_customer)
```

This data shows that out of your total customer base, **92,507 are one-time buyers**, while only **2,913 customers have made repeat purchases**. In percentage terms, repeat customers form a very small portion of your overall audience, which indicates that the business is still heavily dependent on new customer acquisition rather than retention.

From a practical standpoint, this suggests that while your marketing efforts are effective at bringing customers in, they are not as successful at keeping them engaged after their first purchase. However, unlike your previous scenario where repeat customers were zero, this is a slight improvement—it shows that some customers do find value in your products and are willing to come back.

That said, the gap is still quite large. A strong eCommerce business typically relies on a healthy mix of both new and repeat customers, as repeat buyers contribute more consistently to revenue and have a higher lifetime value. Right now, your retention rate appears low, which could impact long-term profitability.

The key opportunity here is to convert more of those one-time buyers into repeat customers. Focusing on strategies like personalized offers, loyalty programs, better post-purchase engagement, and improved customer experience can help bridge this gap and create a more stable and sustainable revenue stream.

RFM-Segementation

```python
data.columns
```

```python
data['order_purchase_timestamp'] = pd.to_datetime(data['order_purchase_timestamp'])
```

```python
import datetime as dt

today = data['order_purchase_timestamp'].max()
rfm = data.groupby('customer_unique_id').agg({
    'order_purchase_timestamp':lambda x: (today -x.max()).days,
    'order_id': 'nunique',
    'payment_value':'sum'
})

rfm.columns = ['Recency','Frequency','Monetary']
rfm.head()
```

This RFM snapshot gives a quick but meaningful view of how a few customers are behaving in your business. All five customers have a frequency of 1, which means they’ve only made a single purchase. That immediately tells you these are one-time buyers, not yet converted into repeat customers.

Looking at recency, there’s a wide spread. One customer purchased relatively recently (around 115–118 days ago), while another hasn’t returned for over 500 days. This variation is important—it shows that some customers are still within a window where re-engagement is possible, while others are at high risk of being completely inactive or churned.

The monetary value also varies significantly. One customer spent nearly ₹197, while another spent just ₹27. This indicates that even among one-time buyers, there are different value segments—some customers have higher potential and are worth targeting more aggressively.

From a business perspective, this highlights a missed opportunity. You have customers who have already made a purchase, but none have been retained. The priority should be to convert these one-time buyers into repeat customers, especially those with higher spending and more recent activity. Personalized offers, reminders, or targeted campaigns could help bring them back and improve overall customer lifetime value.


```python
rfm['R_score'] = pd.qcut(rfm['Recency'], 5, labels=[5,4,3,2,1])
rfm['F_score'] = pd.qcut(rfm['Frequency'].rank(method='first'), 5, labels=[1,2,3,4,5])
rfm['M_score'] = pd.qcut(rfm['Monetary'], 5, labels=[1,2,3,4,5])
```

```python
rfm['RFM_score'] = rfm['R_score'].astype(str) + rfm['F_score'].astype(str) + rfm['M_score'].astype(str)

rfm.head()
```

```python
def segment_customer(row):
    if row['Frequency'] > 5 and row['Monetary'] > 500:
        return 'VIP'
    elif row['Frequency'] > 2:
        return 'Loyal'
    elif row['Recency'] < 100:
        return 'Recent Customers'
    elif row['Recency'] > 300:
        return 'At Risk'
    else:
        return 'Others'

rfm['Segment'] = rfm.apply(segment_customer, axis=1)
```

```python
rfm['Segment'].value_counts()
```

```python
rfm['Segment'].value_counts().plot(kind = 'bar')
plt.title("Customer Segments Distribution")
plt.show()
```

This chart presents a very different and more concerning picture of your customer base. The largest segment here is **“Others”**, followed by a significant number of **“At Risk”** customers, and then **“Recent Customers.”** In contrast, the **“Loyal”** and **“VIP”** segments are almost negligible.

This distribution suggests that most of your customers are either low-engagement or inconsistent buyers who haven’t built a strong relationship with your business. The high number of **“At Risk”** customers is especially important—it indicates that many users who previously interacted with your brand are now becoming inactive. This is often a sign of declining engagement, unmet expectations, or lack of follow-up.

The **“Recent Customers”** segment shows that acquisition is still happening, which is positive. However, the extremely small number of loyal customers implies that new users are not being successfully converted into repeat buyers. Similarly, the absence of a meaningful VIP segment indicates that you currently lack a strong base of high-value, long-term customers.

From a business standpoint, this is a retention problem rather than an acquisition problem. The immediate priority should be to **improve customer experience, build engagement strategies, and introduce retention campaigns**. Without addressing this, the business risks continuously losing customers and relying heavily on acquiring new ones, which is both costly and unsustainable.
